# Question 2: Numerical operations and preprocessing

```python
np.random.seed(101)
scores = np.random.randint(50, 101, size=(15, 3))
means = scores.mean(axis=0).round(2)
stds = scores.std(axis=0)
max_third_exam = scores[:, 2].max()
student_means = scores.mean(axis=1).round(2)
curved_scores = scores.copy()
curved_scores[:, 0] += 5
normalized_scores = (scores - means) / stds

print('scores:\n', scores)
print('means:', means)
print('stds:', stds)
print('max_third_exam:', max_third_exam)
print('student_means:', student_means)
print('curved_scores:\n', curved_scores)
print('normalized means:', normalized_scores.mean(axis=0))
print('normalized stds:', normalized_scores.std(axis=0))
```

# Question3: Data filtering

```python
high_mask = scores > 95
high_count = high_mask.sum()
high_performers_scores = scores[high_mask]
second_exam_mean_for_low_first = scores[scores[:, 0] <= 70, 1].mean().round(2)
pass_fail = scores.mean(axis=1) > 65
excellent_mask = np.all(scores >= 90, axis=1)
excellent_scores = scores[excellent_mask]

print('high_count:', high_count)
print('high_performers_scores:', high_performers_scores)
print('second_exam_mean_for_low_first:', second_exam_mean_for_low_first)
print('pass_fail:', pass_fail)
print('excellent_scores:', excellent_scores if len(excellent_scores) else 'No excellent students found')
```

# Question 4: Exploratory Data Analysis of the Titanic Dataset

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
print('First 5 lines:\n',df.head())
print('info:\n',df.info())
print('describe:\n',df.describe())

survived = df['Survived']
selected_cols = df[['Pclass', 'Sex', 'Age']]
row_3 = df.loc[3]
first_5_3cols = df.iloc[:5, :3]
num_survived = df['Survived'].sum()
survivors = df[df['Survived'] == 1]
num_male_survivors = survivors[survivors['Sex'] == 'male'].shape[0]
first_class_age_over_50 = df[(df['Pclass'] == 1) & (df['Age'] > 50)]

print(type(survived))
print(selected_cols.head())
print(row_3)
print(first_5_3cols)
print(num_survived, num_male_survivors, first_class_age_over_50.shape[0])
```

# Question 5: Data Cleaning and Preprocessing

```python
missing_counts = df.isnull().sum()
df_cleaned = df.drop(columns=['Cabin']).copy()
age_median = df_cleaned['Age'].median()
df_cleaned['Age'] = df_cleaned['Age'].fillna(age_median)
embarked_mode = df_cleaned['Embarked'].mode()[0]
df_cleaned['Embarked'] = df_cleaned['Embarked'].fillna(embarked_mode)
df_cleaned['Pclass'] = df_cleaned['Pclass'].astype('category')

print(missing_counts)
print(df_cleaned.isnull().sum())
```

# Question 6: Feature Engineering and Group Analysis

```python
df_cleaned['FamilySize'] = df_cleaned['SibSp'] + df_cleaned['Parch']
df_cleaned['Title'] = df_cleaned['Name'].str.extract(r',\s*([^\.]+)\.', expand=False).str.strip()
sex_survival = df_cleaned.groupby('Sex')['Survived'].mean()
pclass_survival = df_cleaned.groupby('Pclass')['Survived'].mean()
pclass_sex_survival = df_cleaned.groupby(['Pclass', 'Sex'])['Survived'].mean()

def age_group(x):
    if x < 18:
        return 'Child'
    elif x <= 65:
        return 'Adult'
    return 'Senior'

df_cleaned['AgeGroup'] = df_cleaned['Age'].apply(age_group)
print(df_cleaned[['FamilySize', 'Title', 'AgeGroup']].head())
print(sex_survival)
print(pclass_survival)
print(pclass_sex_survival)
```
